# Figure 2 - lineage marker dot plots

# Load modules

```python
from pathlib import Path
import importlib
import sys

import matplotlib as mpl
import matplotlib.pyplot as plt
import muon as mu
import scanpy as sc
```

# Prepare plots 

```python


plt.rcdefaults()
sc.set_figure_params(vector_friendly=True, dpi_save=300)
mpl.rcParams["pdf.fonttype"] = 42
mpl.rcParams["font.family"] = "sans-serif"
mpl.rcParams["font.sans-serif"] = ["Arial"]
```

# Object and project directory 

```python

PROJECT_DIR = Path("/projects/Pregnancy_placenta")
code_dir = "/projects/Pregnancy_placenta/Code_Publication/Python/"
CODE_DIR = Path(code_dir)
HELPER_FUNCS_DIR = CODE_DIR / "Functions"
MARKER_FILE_PATH = (
    CODE_DIR / "AuxFiles" / "marker genes_v4_PAL.xlsx"
)
FINAL_OBJ_DIR = PROJECT_DIR / "Results" / "21_Final_objects"
EVT_PATH = FINAL_OBJ_DIR / "EVT.h5mu"
FPV_PATH = FINAL_OBJ_DIR / "fPV.h5mu"
FFB_PATH = FINAL_OBJ_DIR / "fFB.h5mu"
ENDO_PATH = PROJECT_DIR / (
    "Results/11_Subclustering_B1-15/Endothelial/PB_RD1_Endo_GEX_0.7/"
    "RD2A_Endo/RD2A_endo_f_webbrowser/RD2A_endoF_web_NoMAT/"
    "RD2A_Endo_f_V5.h5mu"
)
FINAL_FIG_DIR = PROJECT_DIR / "Results" / "22_Final_figures_PAL"
```

# Load functions

```python
for module_dir in (CODE_DIR, HELPER_FUNCS_DIR):
    if str(module_dir) not in sys.path:
        sys.path.append(str(module_dir))

import Func_Marker_dotplot_fig as func_marker_dotplt
importlib.reload(func_marker_dotplt)
```

# Figure 2B - EVT

## Load object and marker definitions

```python
EVT_subcluster = mu.read(str(EVT_PATH))
strip_cat_EVT = func_marker_dotplt.read_markers_dotplot(
    file_path=str(MARKER_FILE_PATH),
    lineage="EVT_main",
)
```

## Plot

```python
func_marker_dotplt.dot_plot_with_umap_stack(
    EVT_subcluster["GEX"],
    prot_anndata=EVT_subcluster["ADT"],
    groupby="sub_lin2",
    strip_categories=strip_cat_EVT,
    shown_categories=["EVT_1", "EVT_2", "EVT_3", "EVT_4", "EVT_5", "EVT_6"],
    legend_fontsize=16,
    legend_fontoutline=0,
    xtick_fontsize=16,
    category_size=16,
    umap_title_fontsize=18,
    legend_length=23,
    umap_title="EVT",
    umap_aspect=1.3,
    legend_dotsize=80,
    standard_scale="var",
    figsize=(13, 4),
    cbar_orientation="vertical",
    cbar_loc=[1.05, 0.3, 0.015, 0.61],
    legend_loc=[0.95, 0.05, 0.2, 0.1],
    save_fig=str(FINAL_FIG_DIR / "Fig2C_EVTsublin_dotplot_V3.pdf"),
    dpi=600,
)
```
