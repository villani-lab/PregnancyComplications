# Sup Fig 5A - EVT

```python
save_dir = FINAL_FIG_DIR
objs_plot = ["EVT1", "EVT3", "EVT5"]

title_overrides = {
    "EVT1": "EVT_1",
    "EVT3": "EVT_2",
    "EVT5": "EVT_5",
}

lineage_title_overrides = {
    "EVT1": "sPTB",
    "EVT3": "sPTB",
    "EVT5": "sPTB",
}

celltype_markers_dict = {
    "EVT1": ["TAC3", "CRABP2"],
    "EVT3": ["ENG", "INHBA"],
    "EVT5": ["COL5A1", "KRT6A"],
}

crop = (1100, 1000, 1300, 1200)

fig, axes = vis_plt.plot_spatial_lineage_markers(
    adata=TMA_obj,
    objs_plot=objs_plot,
    celltype_markers_dict=celltype_markers_dict,
    crop_coord=crop,
    include_cell2location=True,
    marker_columns=2,
    use_raw=False,
    library_id="TMA2-140",
    lineage_title="PTB",
    wspace=0.12,
    hspace=0.1,
    markers_by="row",
    col_title_y=1.06,
    title_overrides=title_overrides,
    lineage_title_overrides=lineage_title_overrides,
    italicize_gene_titles=True,
    scale_bar_um=SCALE_BAR_UM,
    scale_bar_kwargs={
        "bbox_to_anchor": SCALE_BAR_BBOX,
        "size_vertical": LINE_SIZE_VERTICAL,
    },
    save_dir=save_dir,
    save_name="c2l_EVT_spatial_markers_V1",
    dpi=600,
)

plt.show()
```
